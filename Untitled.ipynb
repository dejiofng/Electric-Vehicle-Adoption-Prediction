```python
import pandas as pd
```

```python
df = pd.read_csv("global_ev_adoption_behavior_2026.csv")
df.head()
```

```python
print(df["education_level"].unique())
print(df["city_type"].unique())
print(df["current_vehicle_type"].unique())
print(df["ev_adoption_likelihood"].unique())
```

```python
df.info()
```

```python
pd.set_option("display.max_columns", None)
df.describe()
```

```python
from sklearn.preprocessing import LabelEncoder

for col in df.select_dtypes(include="object").columns:
    encoder = LabelEncoder()
    df[col] = encoder.fit_transform(df[col].astype(str))
```

```python
df.head()
```

```python
df["ev_adoption_likelihood"].value_counts()
```

```python
X = df.drop(["ev_adoption_likelihood"], axis=1)
y = df["ev_adoption_likelihood"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
```

```python
from sklearn.ensemble import RandomForestClassifier
rfc = RandomForestClassifier()
rfc.fit(X_train, y_train)
```

```python
rfc.score(X_test, y_test)
```
